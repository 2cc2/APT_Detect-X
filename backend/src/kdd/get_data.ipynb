```python
import pandas as pd

# Load the original dataset
original_data = pd.read_csv('kddcup.data_10_percent_corrected', header=None)

# Set the random seed for reproducibility
random_seed = 42

# Sample 1% of the data
sampled_data = original_data.sample(frac=0.01, random_state=random_seed)

# Save the sampled data to a new CSV file
sampled_data.to_csv('kddcup_1_percent', index=False, header=False)

print(f'Sampled dataset saved with {len(sampled_data)} entries.')
```
